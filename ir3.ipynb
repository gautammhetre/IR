```python
import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination
```

```python
# Step 2: Load the Heart Disease dataset
url = "https://gist.githubusercontent.com/notreallyme2/8d20f6b5bcc3606e0541cbdf2ee0a7a6/raw/heart.csv"
data = pd.read_csv(url)
```

```python
# Keep only relevant columns
data = data[['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 
             'restecg', 'thalach', 'exang', 'target']]
```

```python
print("Preview of dataset:")
print(data.head())
```

```python
# Step 3: Define the Bayesian Network structure
model = DiscreteBayesianNetwork([
    ('age', 'trestbps'),
    ('age', 'chol'),
    ('cp', 'target'),
    ('sex', 'target'),
    ('trestbps', 'target'),
    ('chol', 'target'),
    ('fbs', 'target'),
    ('exang', 'target')
])
```

```python
# Step 4: Train the model
model.fit(data, estimator=MaximumLikelihoodEstimator)
```

```python
# Step 5: Perform inference
infer = VariableElimination(model)
```

```python
# Step 6: Example queries
print("\n--- Diagnosis Examples ---")
```

```python
# Example 1: Male, 54 years old, chest pain type 2
q1 = infer.query(variables=['target'], evidence={'sex': 1, 'age': 54, 'cp': 2})
print("\nProbability of heart disease for male, 54, cp=2:")
print(q1)
```

```python
# Example 2: Female, 60 years old, exercise-induced angina
q2 = infer.query(variables=['target'], evidence={'sex': 0, 'age': 60, 'exang': 1})
print("\nProbability of heart disease for female, 60, exang=1:")
print(q2)
```
