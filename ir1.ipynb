```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')  # ✅ fix for latest nltk
```

```python
text = "This is a sample sentence showing the effect of removing stopwords and stemming words."
```

```python
# Stopword Removal
words = word_tokenize(text)
filtered = [w for w in words if w.lower() not in stopwords.words('english')]
print("Filtered Text:", filtered)
```

```python
# Stemming
ps = PorterStemmer()
stemmed = [ps.stem(w) for w in filtered]
print("Stemmed Words:", stemmed)
```
